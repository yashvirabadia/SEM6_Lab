```python
import numpy as np
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([0,0,0,1])
```

```python
# weights = np.array([0., 0.0])
# bias = 0.0
# lr = 0.1
# epochs = 5000 
weights = np.array([0.5, 0.7])
bias = 0.9
lr = 0.1
epochs = 500
```

```python
def step_function(z):
    if z >= 0:
        return 1
    return 0
```

```python
for epoch in range(epochs):
    print(f"\nEPOCH -> {epoch+1}")

    total_mistakes = 0

    for i in range(4):
        x = X[i]
        y = Y[i]

        z = np.dot(x, weights) + bias
        pred = step_function(z)
        print(x,y,z,pred)

        error = y - pred

        if error!=0:
            weights = weights + (lr*error+x)
            bias = bias + (lr*error*1)
            total_mistakes += 1
        else:
            print(f"correct {x}")
```
